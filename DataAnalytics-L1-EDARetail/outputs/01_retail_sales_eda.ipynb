## 7. Data Preparation & Validation

```python
df_clean = df.copy()
```

### 7.1 Standardize Column Names

The original dataset contains a typographical error in the `quantiy` column name. It is standardized to `quantity` for readability and consistency.

```python
df_clean = df_clean.rename(columns={'quantiy': 'quantity'})
df_clean.columns.tolist()
```

### 7.2 Convert and Validate Dates

```python
df_clean['sale_date'] = pd.to_datetime(df_clean['sale_date'], errors='coerce')

print('Date type:', df_clean['sale_date'].dtype)
print('Minimum date:', df_clean['sale_date'].min())
print('Maximum date:', df_clean['sale_date'].max())
```

### 7.3 Numerical Validation

```python
numeric_cols = ['age', 'quantity', 'price_per_unit', 'cogs', 'total_sale']
df_clean[numeric_cols].describe().T
```

```python
for col in numeric_cols:
    print(f'\n{col}')
    print('Non-positive values:', (df_clean[col] <= 0).sum())
```

### 7.4 Validate the Total Sales Calculation

For complete observations, total sales should satisfy: **quantity × price per unit = total sale**.

```python
df_clean['calculated_total_sale'] = (
    df_clean['quantity'] * df_clean['price_per_unit']
)

df_clean['sales_difference'] = (
    df_clean['total_sale'] - df_clean['calculated_total_sale']
)

df_clean['sales_difference'].describe()
```

```python
print(
    'Transactions with inconsistent total sales:',
    (df_clean['sales_difference'].abs() > 0.01).sum()
)
```
